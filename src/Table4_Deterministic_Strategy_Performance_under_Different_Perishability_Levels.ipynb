```python
import pandas as pd


a = 200.0
b = 1.0
t = 3.0
C = 40.0
C_w = 2.0
lambda_pen = 2.0
Q = 200.0

theta_disc = 0.00
theta_bund = 0.10
theta_bogo = 0.20

def compute_table3_analytical():
    results = []


    delta_low = 2.0


    P_disc_low = 95.0
    D_disc_low = (1 + theta_disc) * (a - b * P_disc_low - delta_low * t)
    Z_disc_low = P_disc_low * D_disc_low - C * Q - (C_w + lambda_pen) * (Q - D_disc_low)
    results.append({
        "Perishability": "Low (delta = 2.0 < delta*)",
        "Strategy": "Discount",
        "Optimal Parameter": "d* = 0.05",
        "Demand D_s": round(D_disc_low, 1),
        "Objective Value Z_s": round(Z_disc_low, 2),
        "Optimal Choice": ""
    })


    P_bund_low = 75.0
    D_bund_low = (1 + theta_bund) * (a - b * P_bund_low - delta_low * t)
    Z_bund_low = P_bund_low * D_bund_low - C * Q - (C_w + lambda_pen) * (Q - D_bund_low)
    results.append({
        "Perishability": "Low (delta = 2.0 < delta*)",
        "Strategy": "Bundle",
        "Optimal Parameter": "alpha* = 0.75",
        "Demand D_s": round(D_bund_low, 1),
        "Objective Value Z_s": round(Z_bund_low, 2),
        "Optimal Choice": "✓"
    })


    P_bogo_low = 50.0
    D_bogo_low = (1 + theta_bogo) * (a - b * P_bogo_low - delta_low * t)
    Z_bogo_low = P_bogo_low * D_bogo_low - C * Q - (C_w + lambda_pen) * (Q - D_bogo_low)
    results.append({
        "Perishability": "Low (delta = 2.0 < delta*)",
        "Strategy": "BOGO",
        "Optimal Parameter": "Fixed",
        "Demand D_s": round(D_bogo_low, 1),
        "Objective Value Z_s": round(Z_bogo_low, 2),
        "Optimal Choice": ""
    })


    delta_high = 18.0


    P_disc_high = 71.0
    D_disc_high = (1 + theta_disc) * (a - b * P_disc_high - delta_high * t)
    Z_disc_high = P_disc_high * D_disc_high - C * Q - (C_w + lambda_pen) * (Q - D_disc_high)
    results.append({
        "Perishability": "High (delta = 18.0 > delta*)",
        "Strategy": "Discount",
        "Optimal Parameter": "d* = 0.29",
        "Demand D_s": round(D_disc_high, 1),
        "Objective Value Z_s": round(Z_disc_high, 2),
        "Optimal Choice": ""
    })


    P_bund_high = 71.0
    D_bund_high = (1 + theta_bund) * (a - b * P_bund_high - delta_high * t)
    Z_bund_high = P_bund_high * D_bund_high - C * Q - (C_w + lambda_pen) * (Q - D_bund_high)
    results.append({
        "Perishability": "High (delta = 18.0 > delta*)",
        "Strategy": "Bundle",
        "Optimal Parameter": "alpha* = 0.71",
        "Demand D_s": round(D_bund_high, 1),
        "Objective Value Z_s": round(Z_bund_high, 2),
        "Optimal Choice": ""
    })


    P_bogo_high = 50.0
    D_bogo_high = (1 + theta_bogo) * (a - b * P_bogo_high - delta_high * t)
    Z_bogo_high = P_bogo_high * D_bogo_high - C * Q - (C_w + lambda_pen) * (Q - D_bogo_high)
    results.append({
        "Perishability": "High (delta = 18.0 > delta*)",
        "Strategy": "BOGO",
        "Optimal Parameter": "Fixed",
        "Demand D_s": round(D_bogo_high, 1),
        "Objective Value Z_s": round(Z_bogo_high, 2),
        "Optimal Choice": "✓"
    })

    return pd.DataFrame(results)

if __name__ == "__main__":
    df_t3 = compute_table4_analytical()
    print("=== Table 4 Deterministic Strategy Performance under Different Perishability Levels ===")
    print(df_t3.to_string(index=False))


```
